```python
df_clean = df.dropna(axis=1, how="all")
df_clean = df_clean.drop(columns=[
    "STRUCTURE", "STRUCTURE_ID", "STRUCTURE_NAME", "freq", "Time frequency"])
df_clean.shape
```

```python
age_groups = {
    "Y_LT18": "0–17",
    "Y18-24": "18–24",
    "Y25-49": "25–49",
    "Y50-64": "50–64",
    "Y_GE65": "65+"}

df_age = df_clean[df_clean["age"].isin(age_groups)].copy()
df_age["age_group"] = df_age["age"].map(age_groups)
df_age["age_group"].value_counts().sort_index()
```

```python
column_names = {
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "year",
    "statinfo": "indicator",
    "OBS_VALUE": "income",
    "OBS_FLAG": "obs_flag"}

df_clean = df_clean.rename(columns=column_names)
df_age = df_age.rename(columns=column_names)
df_age.head()
```

```python
df_age.info()
```

```python
columns = ["geo_code", "geo_name", "year", "age", "age_group", "sex", "indicator", "unit", "income", "obs_flag"]
df_age = df_age[columns].copy()

df_age["age_group"] = pd.Categorical(
    df_age["age_group"],
    categories=list(age_groups.values()),
    ordered=True)
df_age.info()
```

### Analysis period

The analysis covers survey years 2003–2025. We start in 2003 because it is the first year in the CSV with all five selected age groups. The period ends in 2025, the latest available year. Country coverage and breaks in series are assessed separately before comparisons.

```python
df_clean = df_clean[df_clean["year"].between(2003, 2025)].copy()
df_age = df_age[df_age["year"].between(2003, 2025)].copy()

df_clean.shape, df_age.shape
```

```python
df_clean = df_clean.loc[df_clean["sex"].eq("T")].copy()
df_age = df_age.loc[df_age["sex"].eq("T")].copy()

country_data = df_age.loc[~df_age["geo_code"].str.startswith(("EU", "EA"))].copy()

df_clean.shape, df_age.shape
```

```python
coverage = (
    country_data.groupby(["geo_code", "geo_name"])["year"]
    .agg(
        first_year="min",
        last_year="max",
        years_available="nunique").reset_index())
```

```python
coverage["years_missing"] = 23 - coverage["years_available"]
coverage.sort_values(["years_available", "geo_name"], ascending=[False, True])
```

```python
flag_summary = (
    country_data[country_data["obs_flag"].notna()]
    .groupby(["geo_code", "geo_name", "year", "obs_flag"])
    .size()
    .reset_index(name="observations"))

flag_summary.sort_values(["geo_name", "year"])
```

```python
combination_counts = (
    country_data.groupby(["geo_code", "geo_name", "year"])
    .size()
    .reset_index(name="observations"))

combination_counts["observations"].value_counts().sort_index()
```
