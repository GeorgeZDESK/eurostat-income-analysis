```python
df_clean = df.dropna(axis=1, how="all")
df_clean = df_clean.drop(columns=[
    "STRUCTURE", "STRUCTURE_ID", "STRUCTURE_NAME", "freq", "Time frequency"])
df_clean.shape
```

```python
age_groups = {
    "Y_LT18": "0–17",
    "Y18-24": "18–24",
    "Y25-49": "25–49",
    "Y50-64": "50–64",
    "Y_GE65": "65+"}

df_age = df_clean[df_clean["age"].isin(age_groups)].copy()
df_age["age_group"] = df_age["age"].map(age_groups)
df_age["age_group"].value_counts().sort_index()
```

```python
column_names = {
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "year",
    "statinfo": "indicator",
    "OBS_VALUE": "income",
    "OBS_FLAG": "obs_flag"}

df_clean = df_clean.rename(columns=column_names)
df_age = df_age.rename(columns=column_names)
df_age.head()
```

```python
df_age.info()
```

```python
columns = ["geo_code", "geo_name", "year", "age", "age_group", "sex", "indicator", "unit", "income", "obs_flag"]
df_age = df_age[columns].copy()

df_age["age_group"] = pd.Categorical(
    df_age["age_group"],
    categories=list(age_groups.values()),
    ordered=True)
df_age.info()
```

### Analysis period

The analysis covers survey years 2003–2025. We start in 2003 because it is the first year in the CSV with all five selected age groups. The period ends in 2025, the latest available year. Country coverage and breaks in series are assessed separately before comparisons.

```python
df_clean = df_clean[df_clean["year"].between(2003, 2025)].copy()
df_age = df_age[df_age["year"].between(2003, 2025)].copy()

df_clean.shape, df_age.shape
```

```python
df_clean = df_clean.loc[df_clean["sex"].eq("T")].copy()
df_age = df_age.loc[df_age["sex"].eq("T")].copy()

country_data = df_age.loc[~df_age["geo_code"].str.startswith(("EU", "EA"))].copy()

df_clean.shape, df_age.shape
```

```python
coverage = (
    country_data.groupby(["geo_code", "geo_name"])["year"]
    .agg(
        first_year="min",
        last_year="max",
        years_available="nunique").reset_index())
```

```python
coverage["years_missing"] = 23 - coverage["years_available"]
coverage.sort_values(["years_available", "geo_name"], ascending=[False, True])
```

```python
flag_summary = (
    country_data[country_data["obs_flag"].notna()]
    .groupby(["geo_code", "geo_name", "year", "obs_flag"])
    .size()
    .reset_index(name="observations"))

flag_summary.sort_values(["geo_name", "year"])
```

```python
combination_counts = (
    country_data.groupby(["geo_code", "geo_name", "year"])
    .size()
    .reset_index(name="observations"))

combination_counts["observations"].value_counts().sort_index()
```

```python
key = ["geo_code", "year", "age", "sex", "indicator", "unit"]

print("Duplicate income keys:", df_clean.duplicated(key).sum())
print("Duplicate age-group keys:", df_age.duplicated(key).sum())
```

```python
df_clean.to_csv("../data/processed/ilc_di03_clean.csv", index=False, encoding="utf-8")
```

```python
df_age.to_csv("../data/processed/ilc_di03_age_groups.csv", index=False, encoding="utf-8")
```

### Housing data preparation

```python
housing_raw = pd.read_csv("../data/raw/ilc_lvho07a_linear_2_0.csv")

print("Raw shape:", housing_raw.shape)
display(housing_raw[["STRUCTURE_ID", "freq", "unit"]].drop_duplicates())
```

```python
# Inspect the Greek observations before selecting population groups.
housing_greece = housing_raw.loc[housing_raw["geo"] == "EL"].copy()

print("Available years:", sorted(housing_greece["TIME_PERIOD"].unique()))

display(housing_greece[["rskpovth", "Risk of poverty threshold"]].drop_duplicates())
display(housing_greece[["age", "Age class"]].drop_duplicates())
display(housing_greece[["sex", "Sex"]].drop_duplicates())

print("Column types:")
print(housing_greece[["TIME_PERIOD", "OBS_VALUE", "OBS_FLAG"]].dtypes)

print("Missing values:", housing_greece["OBS_VALUE"].isna().sum())

housing_key = ["freq", "unit", "rskpovth", "age", "sex", "geo", "TIME_PERIOD"]
print("Duplicate observation keys:", housing_greece.duplicated(housing_key).sum())

display(housing_greece[["OBS_FLAG", "Observation status (Flag) V2 structure"]].value_counts(dropna=False))
```

```python
# Identify the flagged observation before filtering by sex.
display(housing_greece.loc[housing_greece["OBS_FLAG"].notna(), ["TIME_PERIOD", "age", "sex", "rskpovth", "OBS_VALUE", "OBS_FLAG"]])
```

```python
# Keep all available age and poverty-status groups for the total population.
housing_clean = housing_greece.loc[
    housing_greece["sex"] == "T",
    [   "geo", "Geopolitical entity (reporting)", "TIME_PERIOD",
        "age", "Age class", "sex",
        "rskpovth", "Risk of poverty threshold",
        "unit", "OBS_VALUE", "OBS_FLAG"]].copy()

housing_clean = housing_clean.rename(columns={
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "housing_overburden_rate",
    "OBS_FLAG": "obs_flag"})

housing_clean = housing_clean.sort_values(["rskpovth", "age", "year"])

print("Prepared shape:", housing_clean.shape)
display(housing_clean.head())
```

```python
# Save the prepared data with observation flags.
housing_clean.to_csv("../data/processed/ilc_lvho07a_greece.csv", index=False, encoding="utf-8")
```

### Income by degree of urbanisation

```python
urban_raw = pd.read_csv("../data/raw/ilc_di17_linear_2_0.csv")

print("Raw shape:", urban_raw.shape)
display(urban_raw[["STRUCTURE_ID", "freq", "unit", "statinfo"]].drop_duplicates())
```

```python
# Inspect the Greek observations before selecting analysis groups.
urban_greece = urban_raw.loc[urban_raw["geo"] == "EL"].copy()

print("Available years:", sorted(urban_greece["TIME_PERIOD"].unique().tolist()))

display(urban_greece[["deg_urb", "Degree of urbanisation"]].drop_duplicates())
display(urban_greece[["age", "Age class", "sex", "Sex"]].drop_duplicates())

print("Column types:")
print(urban_greece[["TIME_PERIOD", "OBS_VALUE", "OBS_FLAG"]].dtypes)

print("Missing values:", urban_greece["OBS_VALUE"].isna().sum())

urban_key = ["freq", "unit", "statinfo", "deg_urb", "sex", "age", "geo", "TIME_PERIOD"]
print("Duplicate observation keys:", urban_greece.duplicated(urban_key).sum())

display(urban_greece[["OBS_FLAG", "Observation status (Flag) V2 structure"]].value_counts(dropna=False))
```

```python
# Keep the published groups and units separate.
urban_clean = urban_greece.loc[
    urban_greece["sex"] == "T",
    [
        "geo", "Geopolitical entity (reporting)", "TIME_PERIOD",
        "age", "Age class", "sex", "statinfo", "unit",
        "deg_urb", "Degree of urbanisation",
        "OBS_VALUE", "OBS_FLAG"
    ]].copy()
```

```python
urban_clean = urban_clean.rename(columns={
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "year",
    "statinfo": "indicator",
    "OBS_VALUE": "income",
    "OBS_FLAG": "obs_flag"})
```

```python
urban_clean = urban_clean.sort_values(["unit", "indicator", "age", "deg_urb", "year"])

print("Prepared shape:", urban_clean.shape)
print("Missing values:", urban_clean["income"].isna().sum())
```

```python
# Identify the retained observations with missing income.
display(urban_clean.loc[urban_clean["income"].isna(), ["year", "age", "deg_urb", "indicator", "unit", "obs_flag"]])
```

```python
# Summarise flagged observations by year.
display(urban_clean.loc[urban_clean["obs_flag"].notna()].groupby(["year", "obs_flag"]).size().reset_index(name="observations"))
```

```python
# Preserve the original values, missingness and flags.
urban_clean.to_csv("../data/processed/ilc_di17_greece.csv", index=False, encoding="utf-8")
```
