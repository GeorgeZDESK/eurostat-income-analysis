### Housing cost overburden across countries

We track the share of people whose household housing costs exceed 40% of disposable income. Both costs and income are net of housing allowances ([Eurostat definition](https://ec.europa.eu/eurostat/statistics-explained/index.php?title=Glossary:Housing_cost_overburden_rate)).

```python
housing_raw = pd.read_csv("../data/raw/ilc_lvho07a_linear_2_0.csv")

# Select the total population, keeping countries rather than EU/EA aggregates.
housing_europe = housing_raw.loc[
    (housing_raw["freq"] == "A")
    & (housing_raw["unit"] == "PC")
    & (housing_raw["age"] == "TOTAL")
    & (housing_raw["sex"] == "T")
    & (housing_raw["rskpovth"] == "TOTAL")
    & housing_raw["TIME_PERIOD"].between(2003, 2025)
    & ~housing_raw["geo"].str.startswith(("EU", "EA"))].copy()

display(
    housing_europe[
        ["STRUCTURE_ID", "freq", "unit", "age", "sex", "rskpovth"]
    ].drop_duplicates())
```

```python
# Summarise coverage without filling values or excluding countries.
housing_coverage = (
    housing_europe.groupby(
        ["geo", "Geopolitical entity (reporting)"]
    )
    .agg(
        first_survey_year=("TIME_PERIOD", "min"),
        last_survey_year=("TIME_PERIOD", "max"),
        survey_years=("TIME_PERIOD", "nunique"),
        rows=("OBS_VALUE", "size"),
        non_missing_values=("OBS_VALUE", "count")
    )
    .reset_index()
    .sort_values("Geopolitical entity (reporting)"))

display(housing_coverage)

print("Missing values:", housing_europe["OBS_VALUE"].isna().sum())
print(
    "Duplicate country-year keys:",
    housing_europe.duplicated(["geo", "TIME_PERIOD"]).sum())
```

```python
# Inspect flags in the selected population before exporting.
display(
    housing_europe.loc[
        housing_europe["OBS_FLAG"].notna(),
        [
            "geo", "TIME_PERIOD", "OBS_VALUE", "OBS_FLAG",
            "Observation status (Flag) V2 structure"
        ]
    ].sort_values(["geo", "TIME_PERIOD"]))
```

```python
housing_europe_clean = housing_europe[
    [
        "geo", "Geopolitical entity (reporting)", "TIME_PERIOD",
        "freq", "unit", "age", "sex", "rskpovth",
        "OBS_VALUE", "OBS_FLAG",
        "Observation status (Flag) V2 structure"]].copy()

housing_europe_clean = housing_europe_clean.rename(columns={
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "housing_overburden_rate",
    "OBS_FLAG": "obs_flag",
    "Observation status (Flag) V2 structure": "flag_description"}).sort_values(["geo_name", "year"])

housing_europe_clean.to_csv("../data/processed/ilc_lvho07a_europe.csv",index=False)

display(housing_europe_clean.head())
```

```python
# Retain both published income groups, all ages and sexes.
# Rates refer to people within each income group.
housing_income_europe = housing_raw.loc[
    (housing_raw["freq"] == "A")
    & (housing_raw["unit"] == "PC")
    & (housing_raw["age"] == "TOTAL")
    & (housing_raw["sex"] == "T")
    & housing_raw["rskpovth"].isin(["B_60", "A_60"])
    & housing_raw["TIME_PERIOD"].between(2003, 2025)
    & ~housing_raw["geo"].str.startswith(("EU", "EA")),
    [
        "geo", "Geopolitical entity (reporting)", "TIME_PERIOD",
        "freq", "unit", "age", "sex",
        "rskpovth", "Risk of poverty threshold",
        "OBS_VALUE", "OBS_FLAG",
        "Observation status (Flag) V2 structure"]].copy()

housing_income_europe = housing_income_europe.rename(columns={
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "year",
    "Risk of poverty threshold": "income_group_label",
    "OBS_VALUE": "housing_overburden_rate",
    "OBS_FLAG": "obs_flag",
    "Observation status (Flag) V2 structure": "flag_description"
}).sort_values(["geo_name", "rskpovth", "year"])

housing_income_key = ["geo_code", "year", "rskpovth"]

print("Prepared shape:", housing_income_europe.shape)
print(
    "Duplicate observation keys:",
    housing_income_europe.duplicated(housing_income_key).sum())
print(
    "Missing values:",
    housing_income_europe["housing_overburden_rate"].isna().sum())

display(
    housing_income_europe[
        ["freq", "unit", "age", "sex", "rskpovth", "income_group_label"]
    ].drop_duplicates().sort_values("rskpovth"))
```

```python
housing_income_2024 = housing_income_europe.loc[housing_income_europe["year"] == 2024].copy()

housing_group_countries = (
    housing_europe_clean[["geo_code", "geo_name"]]
    .drop_duplicates()
    .sort_values("geo_name")
    .set_index("geo_code"))

# Reindex to retain every country and both groups, including absent observations.
housing_group_values_2024 = housing_income_2024.pivot(
    index="geo_code",
    columns="rskpovth",
    values="housing_overburden_rate"
).reindex(
    index=housing_group_countries.index,
    columns=["B_60", "A_60"])

housing_group_coverage_2024 = housing_group_countries.join(
    housing_group_values_2024)

housing_group_coverage_2024["both_available"] = (
    housing_group_coverage_2024[["B_60", "A_60"]]
    .notna()
    .all(axis=1))

print("2024 coverage and housing cost overburden rates (%):")
display(housing_group_coverage_2024)

print("2024 flagged observations:")
display(
    housing_income_2024.loc[
        housing_income_2024["obs_flag"].notna(),
        [
            "geo_code", "geo_name", "rskpovth",
            "housing_overburden_rate", "obs_flag", "flag_description"]])
```

```python
# Preserve all years, missing values and source flags.
housing_income_europe.to_csv("../data/processed/ilc_lvho07a_europe_income_groups.csv", index=False)

print("Saved:", housing_income_europe.shape)
```

### HICP

```python
# Retain EUR and NAC separately for the currency-unit review.
income_for_real = pd.read_csv("../data/processed/ilc_di03_clean.csv")

median_europe_units = income_for_real.loc[
    (income_for_real["age"] == "TOTAL")
    & (income_for_real["sex"] == "T")
    & (income_for_real["indicator"] == "MED_EI")
    & income_for_real["unit"].isin(["EUR", "NAC"])
    & income_for_real["year"].between(2003, 2025)
    & ~income_for_real["geo_code"].str.startswith(("EU", "EA")),
    [
        "geo_code", "geo_name", "year", "unit", "income",
        "obs_flag", "Observation status (Flag) V2 structure"
    ]].copy()

real_country_names = (
    median_europe_units[["geo_code", "geo_name"]]
    .drop_duplicates()
    .set_index("geo_code")["geo_name"]
    .sort_values())

hicp_europe_raw = pd.read_csv("../data/raw/prc_hicp_ainr__custom_23673998_linear_2_0.csv")
```

```python
# Select the annual average all-items index, keeping all available years.
hicp_europe = hicp_europe_raw.loc[
    (hicp_europe_raw["freq"] == "A")
    & (hicp_europe_raw["unit"] == "INX_A_AVG")
    & (hicp_europe_raw["coicop18"] == "TOTAL")
    & hicp_europe_raw["geo"].isin(real_country_names.index),
    [
        "geo", "Geopolitical entity (reporting)", "TIME_PERIOD",
        "freq", "unit", "coicop18", "OBS_VALUE", "OBS_FLAG",
        "Observation status (Flag) V2 structure"
    ]].copy()

hicp_europe = hicp_europe.rename(columns={
    "geo": "geo_code",
    "Geopolitical entity (reporting)": "geo_name",
    "TIME_PERIOD": "price_year",
    "OBS_VALUE": "hicp",
    "OBS_FLAG": "hicp_flag",
    "Observation status (Flag) V2 structure": "hicp_flag_description"
}).sort_values(["geo_name", "price_year"])

print("Prepared HICP shape:", hicp_europe.shape)
print(
    "Duplicate HICP keys:",
    hicp_europe.duplicated(["geo_code", "price_year"]).sum())
print("Missing HICP values:", hicp_europe["hicp"].isna().sum())

print(
    "Duplicate income keys:",
    median_europe_units.duplicated(["geo_code", "year", "unit"]).sum())

display(hicp_europe[["freq", "unit", "coicop18"]].drop_duplicates())
```

```python
median_nac_europe = median_europe_units.loc[median_europe_units["unit"] == "NAC"].copy()

income_coverage = median_nac_europe.groupby("geo_code").agg(
    first_survey_year=("year", "min"),
    last_survey_year=("year", "max"),
    survey_years=("year", "nunique"),
    income_values=("income", "count"))

hicp_coverage = hicp_europe.groupby("geo_code").agg(
    first_price_year=("price_year", "min"),
    last_price_year=("price_year", "max"),
    price_years=("price_year", "nunique"),
    hicp_values=("hicp", "count"))

real_data_coverage = (
    real_country_names.to_frame()
    .join(income_coverage)
    .join(hicp_coverage))

print("Income and HICP coverage:")
display(real_data_coverage)

print("Flagged HICP observations:")
display(
    hicp_europe.loc[
        hicp_europe["hicp_flag"].notna(),
        [
            "geo_name", "price_year", "hicp",
            "hicp_flag", "hicp_flag_description"
        ]])

print("Flagged NAC income observations:")
display(
    median_nac_europe.loc[
        median_nac_europe["obs_flag"].notna(),
        [
            "geo_name", "year", "income", "obs_flag",
            "Observation status (Flag) V2 structure"
        ]
    ].sort_values(["geo_name", "year"]))
```

```python
currency_unit_check = median_europe_units.pivot(
    index=["geo_code", "geo_name", "year"],
    columns="unit",
    values="income"
).reindex(columns=["EUR", "NAC"])

# Compare only records with both published values available.
currency_unit_check = (
    currency_unit_check
    .dropna(subset=["EUR", "NAC"])
    .reset_index())

currency_unit_check["nac_equals_eur"] = (currency_unit_check["NAC"] == currency_unit_check["EUR"])

currency_unit_summary = (
    currency_unit_check
    .groupby(["geo_code", "geo_name", "nac_equals_eur"])
    .agg(
        first_survey_year=("year", "min"),
        last_survey_year=("year", "max"),
        observations=("year", "nunique")
    )
    .reset_index()
    .sort_values(["geo_name", "first_survey_year"]))

display(currency_unit_summary)
```

```python
# Preserve all available years, missing values and source flags.
hicp_europe.to_csv("../data/processed/hicp_europe.csv", index=False)

print("Saved:", hicp_europe.shape)
```

```python
# This preparation is specific to income years 2019–2023.
real_income_input = median_europe_units.loc[
    (median_europe_units["unit"] == "NAC")
    & median_europe_units["year"].between(2020, 2024)].copy()

real_income_input = real_income_input.rename(columns={
    "year": "survey_year",
    "income": "income_nac",
    "obs_flag": "income_flag",
    "Observation status (Flag) V2 structure": "income_flag_description"})
```

```python
# These survey years use the previous calendar year as the income reference period.
real_income_input["income_year"] = (real_income_input["survey_year"] - 1)
real_income_input["currency_divisor"] = 1.0
real_income_input["analysis_unit"] = "NAC"
```

```python
# In the original export, Croatian NAC values use kuna through survey year 2022.
# Fixed conversion source:
# https://www.ecb.europa.eu/euro/changeover/croatia/html/index.en.html
legacy_kuna = ((real_income_input["geo_code"] == "HR") & (real_income_input["survey_year"] <= 2022))

real_income_input.loc[legacy_kuna, "currency_divisor"] = 7.53450
real_income_input.loc[real_income_input["geo_code"] == "HR", "analysis_unit"] = "EUR"

real_income_input["income_for_deflation"] = (real_income_input["income_nac"] / real_income_input["currency_divisor"])

print("Croatian currency adjustment:")
display(
    real_income_input.loc[
        real_income_input["geo_code"] == "HR",
        [
            "survey_year", "income_year", "income_nac",
            "currency_divisor", "income_for_deflation", "analysis_unit"
        ]].sort_values("income_year"))

real_income_input.to_csv("../data/processed/ilc_di03_real_input_2019_2023.csv", index=False)

print("Saved:", real_income_input.shape)
```
